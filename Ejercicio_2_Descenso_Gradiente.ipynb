## Datos del problema

Función original:

$$
f(x)=\frac{1}{2}x^TAx-b^Tx
$$

Matriz A:

$$
A=\begin{pmatrix}
1&0&0\\
0&3&0\\
0&0&5
\end{pmatrix}
$$

Vector b:

$$
b=\begin{pmatrix}
-3\\
6\\
-10
\end{pmatrix}
$$

Tasa de aprendizaje:

$$
\gamma=0.25
$$

Punto inicial:

$$
x_0=\begin{pmatrix}
1\\
0\\
1
\end{pmatrix}
$$

## Objetivo

Implementar el descenso del gradiente, comprobar las primeras tres iteraciones realizadas a mano y continuar hasta alcanzar el mínimo.


```python

import numpy as np
import pandas as pd
```

```python

# Matriz A
A = np.array([[1, 0, 0],
              [0, 3, 0],
              [0, 0, 5]])

# Vector b
b = np.array([-3, 6, -10])

# Tasa de aprendizaje
gamma = 0.25

# Punto inicial
x0 = np.array([1.0, 0.0, 1.0])

# Mostrar los datos
print("Matriz A:")
print(A)

print("Vector b:", b)
print("Tasa de aprendizaje:", gamma)
print("Punto inicial:", x0)
```


## Implementación de la función

Se utiliza la función cuadrática del ejercicio:

$$
f(x)=\frac{1}{2}x^TAx-b^Tx
$$

Su expresión desarrollada es:

$$
f(x_1,x_2,x_3)=\frac{1}{2}x_1^2+\frac{3}{2}x_2^2+\frac{5}{2}x_3^2+3x_1-6x_2+10x_3
$$


```python

# Función matemática del ejercicio 2
def f(x):
    return 0.5 * (x.T @ A @ x) - (b.T @ x)

# Evaluar la función en el punto inicial
resultado = f(x0)

print("f(x0) =", resultado)
```


## Implementación del gradiente

El gradiente de la función se calcula mediante:

$$
\nabla f(x)=Ax-b
$$

Su expresión desarrollada es:

$$
\nabla f(x)=
\begin{pmatrix}
x_1+3\\
3x_2-6\\
5x_3+10
\end{pmatrix}
$$


```python

# Gradiente de la función
def gradiente(x):
    return A @ x - b

# Evaluar el gradiente en el punto inicial
print("Gradiente en x0:", gradiente(x0))
```


## Descenso del gradiente

Se implementa el método de descenso del gradiente mediante la fórmula:

$$
x_{i+1}=x_i-\gamma\nabla f(x_i)
$$

Se calculan las primeras tres iteraciones para comprobar que los resultados obtenidos en Python coinciden con los desarrollados a mano.


```python

# Inicializar el punto actual
x = x0.copy()

# Lista para almacenar resultados
resultados = []

# Calcular desde x0 hasta x3
for i in range(4):

    # Calcular el gradiente
    g = gradiente(x)

    # Evaluar la función
    valor = f(x)

    # Guardar los resultados
    resultados.append({
        "Iteración": i,
        "x1": x[0],
        "x2": x[1],
        "x3": x[2],
        "Gradiente x1": g[0],
        "Gradiente x2": g[1],
        "Gradiente x3": g[2],
        "f(x)": valor
    })

    # Actualizar el punto hasta x3
    if i < 3:
        x = x - gamma * g

# Crear la tabla
tabla = pd.DataFrame(resultados)

# Mostrar la tabla
display(tabla)
```

```python

# Valores calculados a mano
valores_manual = np.array([
    [1.0, 0.0, 1.0],
    [0.0, 1.5, -2.75],
    [-0.75, 1.875, -1.8125],
    [-1.3125, 1.96875, -2.046875]
])

# Valores obtenidos en Python
valores_python = tabla[["x1", "x2", "x3"]].to_numpy()

# Comprobar coincidencia
coinciden = np.allclose(valores_python, valores_manual)

print("¿Coinciden las iteraciones con la libreta?", coinciden)
```


## Convergencia del descenso del gradiente

Se continúa aplicando el descenso del gradiente hasta alcanzar el criterio de convergencia:

$$
\|\nabla f(x_i)\| < 10^{-6}
$$

Se utiliza la tasa de aprendizaje $\gamma=0.25$ y el punto inicial $x_0=(1,0,1)$.

Se reportará el punto mínimo encontrado, el valor de la función, el gradiente final y el número de iteraciones.


```python

# Parámetros del descenso del gradiente
x = x0.copy()
tolerancia = 1e-6
max_iteraciones = 10000

# Contador
iteraciones = 0

# Descenso del gradiente
while (np.linalg.norm(gradiente(x)) >= tolerancia
       and iteraciones < max_iteraciones):

    g = gradiente(x)

    x = x - gamma * g

    iteraciones += 1

# Mostrar resultados
print("Punto mínimo encontrado:", x)
print("Valor mínimo de la función:", f(x))
print("Gradiente final:", gradiente(x))
print("Norma del gradiente:", np.linalg.norm(gradiente(x)))
print("Número de iteraciones:", iteraciones)

print("¿Se alcanzó la convergencia?",
      np.linalg.norm(gradiente(x)) < tolerancia)
```

```python

# Mínimo obtenido analíticamente
minimo_analitico = np.array([-3.0, 2.0, -2.0])

# Comparación de resultados
print("Mínimo analítico:", minimo_analitico)
print("Mínimo obtenido en Python:", x)

print("¿Coinciden aproximadamente?",
      np.allclose(x, minimo_analitico, atol=1e-6))
```
