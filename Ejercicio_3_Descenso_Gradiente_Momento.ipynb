
# Ejercicio 3: Descenso del gradiente con momento

**Materia:** Matemáticas para Inteligencia Artificial

## Datos del problema

Función original:

$$
f(x)=\frac{1}{4}x^4-2x^2
$$

Tasa de aprendizaje:

$$
\gamma=0.05
$$

Momento:

$$
\alpha=0.5
$$

Punto inicial:

$$
x_0=1
$$

## Objetivo

Implementar el descenso del gradiente con momento, comprobar las primeras tres iteraciones realizadas a mano y continuar hasta alcanzar el mínimo de la función.

## Fórmula del método

$$
x_{i+1}=x_i-\gamma f'(x_i)+\alpha\Delta x_i
$$

Donde:

$$
\Delta x_i=x_i-x_{i-1}
$$

Para la primera iteración:

$$
\Delta x_0=0
$$


```python

import numpy as np
import pandas as pd
```

```python

# Tasa de aprendizaje
gamma = 0.05

# Coeficiente de momento
alpha = 0.5

# Punto inicial
x0 = 1.0

# Desplazamiento inicial
delta_x0 = 0.0

# Mostrar los parámetros
print("Tasa de aprendizaje:", gamma)
print("Momento:", alpha)
print("Punto inicial:", x0)
print("Desplazamiento inicial:", delta_x0)
```


## Implementación de la función

Se utiliza la función proporcionada en el ejercicio:

$$
f(x)=\frac{1}{4}x^4-2x^2
$$

Esta función será utilizada para calcular el valor de cada iteración del método de descenso del gradiente con momento.


```python

# Función matemática del ejercicio 3
def f(x):
    return 0.25 * x**4 - 2 * x**2

# Evaluar en el punto inicial
resultado = f(x0)

print("f(x0) =", resultado)
```


## Implementación de la primera derivada

La primera derivada de la función es:

$$
f'(x)=x^3-4x
$$

Esta derivada se utilizará para calcular las actualizaciones del descenso del gradiente con momento.


```python

# Primera derivada
def derivada(x):
    return x**3 - 4*x

# Evaluar la derivada en el punto inicial
print("Derivada en x0:", derivada(x0))
```


## Descenso del gradiente con momento

Se implementa el método utilizando:

$$
x_{i+1}=x_i-\gamma f'(x_i)+\alpha\Delta x_i
$$

Donde:

$$
\Delta x_i=x_i-x_{i-1}
$$

Se utiliza un desplazamiento inicial de cero.

Se calculan las primeras tres iteraciones para comprobar los resultados desarrollados a mano.


```python

# Valores iniciales
x = x0
delta_x = delta_x0

# Lista para guardar resultados
resultados = []

# Calcular desde la iteración 0 hasta la 3
for i in range(4):

    # Calcular derivada y función
    d = derivada(x)
    valor = f(x)

    # Guardar los resultados
    resultados.append({
        "Iteración": i,
        "x_i": x,
        "Delta x_i": delta_x,
        "f'(x_i)": d,
        "f(x_i)": valor
    })

    # Actualizar el punto usando momento
    if i < 3:
        nuevo_x = x - gamma * d + alpha * delta_x

        # Calcular nuevo desplazamiento
        delta_x = nuevo_x - x

        # Actualizar el punto
        x = nuevo_x

# Crear tabla
tabla = pd.DataFrame(resultados)

# Mostrar resultados
display(tabla)
```

```python

# Valores obtenidos a mano
valores_manual = np.array([
    1.0,
    1.15,
    1.37895625,
    1.6381199572
])

# Valores calculados en Python
valores_python = tabla["x_i"].to_numpy()

# Comparación
coinciden = np.allclose(
    valores_python,
    valores_manual,
    atol=1e-6
)

print("¿Coinciden las iteraciones con la libreta?", coinciden)
```


## Convergencia del descenso del gradiente con momento

Se continúa aplicando el descenso del gradiente con momento hasta alcanzar el criterio de convergencia:

$$
|f'(x_i)| < 10^{-6}
$$

Se utilizan los parámetros:

- Tasa de aprendizaje: $\gamma=0.05$
- Momento: $\alpha=0.5$
- Punto inicial: $x_0=1$

Se reporta el mínimo encontrado, el valor de la función, la derivada final y el número de iteraciones.


```python

# Parámetros del algoritmo
x = x0
delta_x = delta_x0

tolerancia = 1e-6
max_iteraciones = 10000
iteraciones = 0

# Descenso del gradiente con momento
while abs(derivada(x)) >= tolerancia and iteraciones < max_iteraciones:

    d = derivada(x)

    nuevo_x = x - gamma * d + alpha * delta_x

    delta_x = nuevo_x - x

    x = nuevo_x

    iteraciones += 1

# Mostrar resultados
print("Mínimo encontrado:", x)
print("Valor de la función:", f(x))
print("Derivada final:", derivada(x))
print("Valor absoluto de la derivada:", abs(derivada(x)))
print("Número de iteraciones:", iteraciones)

print("¿Se alcanzó la convergencia?",
      abs(derivada(x)) < tolerancia)
```

```python

# Mínimo analítico esperado
minimo_analitico = 2.0

print("Mínimo analítico:", minimo_analitico)
print("Mínimo obtenido en Python:", x)

print("¿Coinciden aproximadamente?",
      np.isclose(x, minimo_analitico, atol=1e-6))
```
