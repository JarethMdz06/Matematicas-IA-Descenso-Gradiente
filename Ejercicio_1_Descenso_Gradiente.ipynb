## Datos del problema

Función original:

$$
f(x)=\frac{1}{2}x^TAx-b^Tx
$$

Matriz A:

$$
A=\begin{pmatrix}
4 & 0\\
0 & 10
\end{pmatrix}
$$

Vector b:

$$
b=\begin{pmatrix}
-8\\
10
\end{pmatrix}
$$

Tasa de aprendizaje:

$$
\gamma=0.15
$$

Punto inicial:

$$
x_0=\begin{pmatrix}
0\\
0
\end{pmatrix}
$$


```python

import numpy as np
import pandas as pd
```

```python

# Matriz A
A = np.array([[4, 0],
              [0, 10]])

# Vector b
b = np.array([-8, 10])

# Tasa de aprendizaje
gamma = 0.15

# Punto inicial
x0 = np.array([0.0, 0.0])

# Mostrar los datos
print("Matriz A:")
print(A)

print("Vector b:", b)
print("Tasa de aprendizaje:", gamma)
print("Punto inicial:", x0)
```


## Implementación de la función

Se utiliza la función cuadrática proporcionada en el ejercicio:

$$
f(x)=\frac{1}{2}x^TAx-b^Tx
$$

Su expresión desarrollada es:

$$
f(x_1,x_2)=2x_1^2+5x_2^2+8x_1-10x_2
$$


```python

# Función matemática del ejercicio 1
def f(x):
    return 0.5 * (x.T @ A @ x) - (b.T @ x)

# Evaluar la función en el punto inicial
resultado = f(x0)

print("f(x0) =", resultado)
```

```python

# Gradiente de la función
def gradiente(x):
    return A @ x - b

# Evaluar el gradiente en el punto inicial
print("Gradiente en x0:", gradiente(x0))
```


## Descenso del gradiente

Se implementa el método de descenso del gradiente utilizando:

$$
x_{i+1}=x_i-\gamma\nabla f(x_i)
$$

Se calcularán las primeras tres iteraciones para comprobar que coincidan con los resultados obtenidos a mano.


```python

# Inicializar el punto actual
x = x0.copy()

# Lista para almacenar los resultados
resultados = []

# Realizar tres iteraciones
for i in range(4):

    # Calcular el gradiente
    g = gradiente(x)

    # Calcular el valor de la función
    valor = f(x)

    # Guardar los resultados
    resultados.append({
        "Iteración": i,
        "x1": x[0],
        "x2": x[1],
        "Gradiente x1": g[0],
        "Gradiente x2": g[1],
        "f(x)": valor
    })

    # Actualizar el punto (solo hasta x3)
    if i < 3:
        x = x - gamma * g

# Crear tabla
tabla = pd.DataFrame(resultados)

# Mostrar tabla
display(tabla)
```


## Convergencia del descenso del gradiente

Se continúa aplicando el método de descenso del gradiente hasta alcanzar el mínimo de la función.

El criterio de convergencia utilizado es:

$$
\|\nabla f(x_i)\| < 10^{-6}
$$

Se utiliza una tasa de aprendizaje de $\gamma=0.15$ y el punto inicial $x_0=(0,0)$.


```python

# Parámetros del descenso del gradiente
x = x0.copy()
tolerancia = 1e-6
max_iteraciones = 10000

# Contador de iteraciones
iteraciones = 0

# Descenso del gradiente
while np.linalg.norm(gradiente(x)) >= tolerancia and iteraciones < max_iteraciones:

    g = gradiente(x)

    x = x - gamma * g

    iteraciones += 1

# Mostrar resultados
print("Punto mínimo encontrado:", x)
print("Valor mínimo de la función:", f(x))
print("Gradiente final:", gradiente(x))
print("Norma del gradiente:", np.linalg.norm(gradiente(x)))
print("Número de iteraciones:", iteraciones)
```

```python

# Mínimo obtenido analíticamente
minimo_analitico = np.array([-2.0, 1.0])

# Comparación con el resultado numérico
print("Mínimo analítico:", minimo_analitico)
print("Mínimo obtenido con Python:", x)

print("¿Coinciden aproximadamente?",
      np.allclose(x, minimo_analitico, atol=1e-6))
```
